```
gene_data = {}

while True:
    gene_id = input("Enter Gene ID (or type 'done' to finish): ")
    if gene_id.lower() == 'done':
        break

    sequence = input(f"Enter the DNA sequence for {gene_id}: ")

    # This line "inputs" the pair into your dictionary
    gene_data[gene_id] = sequence

print("\nFinal Dictionary Created:")
print(gene_data)
def analyze_sequence(gene_data):
    results = {}

    for gene_id,gene_sequence in gene_data.items():
      gene_sequence = gene_sequence.upper()
      gene_length = len(gene_sequence)
      if gene_length >= 10:
        if gene_sequence.startswith("ATG"):
          G_count = gene_sequence.count('G')
          C_count = gene_sequence.count('C')

          GC_content = ((G_count+C_count)/gene_length) *100
          results[gene_id] =round(GC_content,2)
        else:
          print("skipping {gene_id}: no starting sequence ATG found")

    return results
print(analyze_sequence(gene_data))
```

```
import time
from functools import wraps

def timer_decorator(func):
  @wraps(func)
  def wrapper(*args, **kwargs):
    start = time.time()
    result = func(*args,**kwargs)
    end = time.time()
    print(f"function{func.__name__} took {end - start:.4f} seconds")
    return result
  return wrapper # Added: The decorator must return the wrapper function

@timer_decorator
def heavy_computation(n):
  return sum(i * i for i in range(n))

heavy_computation(1000000)
```

```
from Bio.Seq import Seq

my_seq = Seq("GATGGTACCGTGCGC")
add_seq = Seq("ATGC")
print(f"Sequence:{my_seq}")
## reverse of sequence is complement
print("reverse or complement sequence =  :{my_seq.reverse_complement()}")
## from dna sequence to get mrna we transcribe
mrna = my_seq.transcribe()
print(f"transcripted succesfully - mrna = :{mrna}")
## to get protien from mrna we translate it
protien = mrna.translate()
print(f"translated protien sequence  = :{protien}")
## the Seq object can be sliced
print("slice = ", my_seq[0:5],"length =", len(my_seq), "add =", my_seq + add_seq)


## GC content calculation
from Bio.SeqUtils import gc_fraction

content = gc_fraction(my_seq)
print(f"gc content: {content * 100:.2f}%")
```

```
## writing files using SeqIO.write()
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction

high_gc = [] # to store higc seq

for record in SeqIO.parse("practice.fasta", "fasta"):
  if gc_fraction(record)>0.50:
    high_gc.append(record)


with open("high_gc.fasta", "w") as output_handle:
  SeqIO.write(high_gc, output_handle, "fasta")

print(f"saved{len(high_gc)} sequences to high_gc.fasta")
print (high_gc)
```
